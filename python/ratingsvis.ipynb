```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.colors import LinearSegmentedColormap
```

```python
# Load the data
df = pd.read_csv('rating_distribution.csv')

# Sort ratings by frequency
df = df.sort_values(by='total', ascending=True)
```

```python
# Create Netflix gradient
netflix_cmap = LinearSegmentedColormap.from_list("netflix_gradient", ["#141414", "#e30913"])
colors = [netflix_cmap(i/len(df)) for i in range(len(df))]

# Plot
plt.figure(figsize=(10, 8))
sns.barplot(
    x='total',
    y='rating',
    data=df,
    hue='rating',      # Suppress palette warning
    dodge=False,
    palette=colors,
    legend=False
)

plt.title('Rating Distribution on Netflix')
plt.xlabel('Number of Titles')
plt.ylabel('Rating')
plt.tight_layout()
```

```python
plt.savefig('visualizations/rating_distribution.png', dpi=300)
plt.show()
```
