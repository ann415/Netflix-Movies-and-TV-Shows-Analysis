```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the data
df = pd.read_csv('titles_by_year.csv')
```

```python
# Sort by year
df = df.sort_values(by='release_year')

# Plotting
plt.figure(figsize=(12, 6))
sns.lineplot(
    data=df,
    x='release_year',
    y='totalNo',
    marker='o',
    linewidth=2,
    color='#e30913'  # Netflix red
)

plt.title('Netflix Content Release Trend by Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
```

```python
plt.savefig('visualizations/content_trend_by_year.png', dpi=300)
plt.show()
```
