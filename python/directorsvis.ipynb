```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.colors import LinearSegmentedColormap
import numpy as np
```

```python
# Load dataset
df = pd.read_csv('top_10_directors.csv')
```

```python
# Sort directors by count
df = df.sort_values(by='frequency', ascending=True)  # for horizontal bar chart
```

```python
# Create Netflix gradient (red to dark)
netflix_cmap = LinearSegmentedColormap.from_list("netflix_gradient", ["#e30913", "#141414"])
colors = [netflix_cmap(i/9) for i in range(10)]  # Generate 10 gradient colors

# Plot
plt.figure(figsize=(10,6))
sns.barplot(
    x='frequency',
    y='director',
    data=df,
    hue='director',          # Assigning hue to suppress the warning
    dodge=False,             # Prevent bars from splitting due to hue
    palette=colors,
    legend=False             # Suppressed legend to avoid warning
)
plt.title('Top 10 Most Frequent Directors on Netflix')
plt.xlabel('Number of Titles')
plt.ylabel('Director')
plt.tight_layout()
```

```python
plt.savefig('visualizations/top_10_directors.png', dpi=300)
plt.show()
```
