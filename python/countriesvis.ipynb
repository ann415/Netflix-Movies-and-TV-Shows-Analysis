```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.colors import LinearSegmentedColormap
```

```python
# Load your dataset
df = pd.read_csv('top_20_countries.csv')
```

```python
# Sort for horizontal bar chart
df = df.sort_values(by='totalNo', ascending=True)
```

```python
# Create Netflix gradient palette
netflix_cmap = LinearSegmentedColormap.from_list("netflix_gradient", ["#141414", "#e30913"])
colors = [netflix_cmap(i/19) for i in range(20)]  # Gradient for top 20

# Plot
plt.figure(figsize=(10, 8))
sns.barplot(
    x='totalNo',
    y='country',
    data=df,
    hue='country',      # Suppress palette warning
    dodge=False,
    palette=colors,
    legend=False
)

# Titles and labels
plt.title('Top 20 Countries Producing Netflix Titles')
plt.xlabel('Number of Titles')
plt.ylabel('Country')
plt.tight_layout()
```

```python
# Save and show
plt.savefig('visualizations/top_20_countries.png', dpi=300)
plt.show()
```
